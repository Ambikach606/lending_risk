##  Read Source Data


```python
try:
    raw_df = (
        spark.read
        .option("header", True)
        .option("inferSchema", False)
        .option("sep", ",")
        .csv(str(INPUT_FILE))
    )

    print(f"Raw record count: {raw_df.count()}")

    raw_df.show(truncate=False)

except FileNotFoundError as exc:
    raise FileNotFoundError(
        f"Input file not found: {INPUT_FILE}"
    ) from exc

except Exception as exc:
    raise RuntimeError(
        f"Failed to read or display input data: {exc}"
    ) from exc
```

##  Execute Transformations : Transform, validate, deduplicate and Show the report summary


```python
try:
    detail_df, rejected_df, duplicates_df = transform_data(raw_df)

    summary_df = make_summary(detail_df)
    print(f"Valid detail rows      : {detail_df.count()}")
    print(f"Rejected rows          : {rejected_df.count()}")
    print(f"Superseded duplicates  : {duplicates_df.count()}")
    print(f"Summary   : {summary_df.count()}")
    print("Detail Data")
    detail_df.show(truncate=False)
    print("Rejected Data")
    rejected_df.select(
        "facility_id",
        "actual_value",
        "rejection_reason"
    ).show(truncate=False)
    print("Duplicate Data")
    duplicates_df.show(truncate=False)
    print("Summary Data")
    summary_df.show(truncate=False)

except ValueError as exc:
    print(f"Input validation error: {exc}")

except RuntimeError as exc:
    print(f"Data processing error: {exc}")
    raise
```

## Export CSV files for Power BI
Writes a folder containing CSV part files for each output.

```python
save_spark_df_to_csv(
    raw_df,
    OUTPUT_DIR / "raw.csv"
)

save_spark_df_to_csv(
    detail_df,
    OUTPUT_DIR / "detail.csv"
)

save_spark_df_to_csv(
    rejected_df,
    OUTPUT_DIR / "rejected.csv"
)

save_spark_df_to_csv(
    duplicates_df,
    OUTPUT_DIR / "duplicates.csv"
)

save_spark_df_to_csv(
    summary_df,
    OUTPUT_DIR / "summary.csv"
)

print(
    f"All output files successfully saved to: {OUTPUT_DIR}"
)
```
