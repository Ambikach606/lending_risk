##  Helper Functions


```python
import warnings
def save_spark_df_to_csv(df, file_path):
    """
    Save a Spark DataFrame as a single CSV file.
    Suitable for small-to-medium datasets.
    """

    df.toPandas().to_csv(
            file_path,
            index=False,
            encoding="utf-8"
        )

    print(f"Saved: {file_path}")
```
